```python
import numpy as np                                       # 다차원 배열(Matrix) 처리 및 수학 연산을 위한 라이브러리
import pandas as pd                                      # 데이터프레임(표 형태의 데이터) 조작 및 분석을 위한 라이브러리
import matplotlib.pyplot as plt                          # 학습 결과(Loss, Accuracy) 및 그래프 시각화를 위한 라이브러리

from sklearn.model_selection import train_test_split     # 전체 데이터를 학습용(Train)과 검증용(Test) 데이터셋으로 분리하는 함수
from sklearn.preprocessing import StandardScaler         # 데이터의 피처(Feature)들을 평균 0, 표준편차 1로 만드는 표준화(정규화) 전처리 클래스

# --- 모델 성능 평가 지표 (Metrics) ---
from sklearn.metrics import accuracy_score               # 정확도(정답을 맞춘 비율) 측정 함수
from sklearn.metrics import confusion_matrix             # 혼동 행렬(정답과 오답의 분포를 표로 표현) 생성 함수
from sklearn.metrics import classification_report        # 정밀도(Precision), 재현율(Recall), F1-Score 등 종합 평가지표 리포트 출력 함수
from sklearn.metrics import roc_auc_score                # 이진 분류 모델의 예측 성능을 정량화하는 ROC-AUC 점수 계산 함수
from sklearn.metrics import auc                          # 곡선 아래 면적(Area Under the Curve)을 계산하는 함수 (주로 ROC 곡선 면적 계산 시 사용)

from sklearn.datasets import load_breast_cancer          # 사이킷런에서 제공하는 유방암 진단 데이터셋 로드 함수

import joblib                                            # 완성된 머신러닝 모델이나 전처리 객체(StandardScaler 등)를 파일로 저장하고 불러오는 라이브러리

# --- 텐서플로우/케라스 인공신경망 구축 ---
from tensorflow.keras import Sequential                  # 레이어를 순차적으로 쌓아서 딥러닝 모델을 만드는 컨테이너(뼈대)
from tensorflow.keras.layers import Dense, Dropout       # Dense: 전결합층(Fully Connected Layer), Dropout: 과적합 방지를 위해 무작위로 뉴런을 끄는 레이어
from tensorflow.keras.callbacks import EarlyStopping     # 학습 중 검증 손실(Loss)이 더이상 줄어들지 않으면 학습을 자동으로 조기 종료하는 콜백 함수
```
