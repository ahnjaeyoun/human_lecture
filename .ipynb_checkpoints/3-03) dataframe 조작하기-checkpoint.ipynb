```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(np.random.randn(6,4))
df
```

```python
df.columns = ['a','b','c','d']
df.index = pd.date_range('20250501', periods=6)
```

```python
df.index
```

```python
df
```

```python
df['f'] = [1.0, np.nan, 3.5, 6.1, np.nan, 7.0]
```

```python
df
```

```python
# NAN 제거하기
df.dropna(how='any')  # nan이 하나라도 포함된 경우 그 행을 삭제
```

```python
df.dropna(how='all')  # nan이 행 전체에 있을 경우, 그 행을 삭제한다.
```

```python
# nan 값 대체하기
df.fillna(value=5.0)
```

```python
# 불리언 마스크로 값을 제어하기
df.isnull()
```

```python
# nan이 있는 행 얻어오기
df.loc[df.isnull()['f'], :]
```

```python
# 인덱스 기준으로 행삭제
pd.to_datetime('20250505')
```

```python
df.drop(pd.to_datetime('20250505'))
```

```python
# 인덱스 기준으로 두 개 이상의 행 삭제하기
df.drop([pd.to_datetime('20250504'), pd.to_datetime('20250506')])
```

```python
# 열 삭제하기
df.drop('f', axis=1)
```

```python
df.drop(['b', 'd'], axis=1)
```
