# 2) DataFrame

```python
data = {"names" : ['Hyunsuk', 'Hyunsuk', 'Hyunsuk', 'Charles', 'Charles'],
       "year" : [2014, 2015, 2016, 2015, 2016],
       "points" : [1.5, 1.7, 3.6, 2.4, 2.9]}
df = pd.DataFrame(data)
df
```

```python
df.shape
```

```python
df.index
```

```python
df.columns
```

```python
df.values
```

```python
# describe 함수
'''
DataFrame df에 대하여 df.describe() 함수를 실행하게 되면,
계산이 가능한 컬럼에 한해서 각 컬럼의 평균, 분산, 최솟/최댓값
'''
```

```python
df.describe()
```
