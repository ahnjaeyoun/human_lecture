# numpy ndarray
- numpy는 특히 벡터 및 행렬 연산에 있어 편의성을 제공하는 라이브러리
- numpy에서는 기본적으로 array(어레이)라는 단위로 데이터를 관리하고, 이에 대한 연산을 수행

# 1차원 array

```python
import numpy as np
```

```python
data = [6,7.5,8,0,1]
```

```python
arr1 = np.array(data)
arr1
```

```python
arr1.shape
```

```python
arr = np.array([2,3,4])
arr
```

# 2차원 array

```python
data2 = [[1,2,3,4],
         [5,6,7,8]]
data2
```

```python
arr2 = np.array(data2)
arr2
```

```python
arr2.shape
```

# zeros(), ones(), arange() 함수

```python
# zeros는 0으로 된 array를 만든다
np.zeros((3,6))
```

```python
# ones는 1로된 array를 만든다
np.ones(10)
```

```python
# arange는 연속된 수로 array를 만든다
np.arange(15)
```

```python
arr1.dtype
```

```python
arr2.dtype
```

```python
# 데이터 타입을 직접 지정할 수 있음
arr = np.array([1,2,3,4,5], dtype=np.int64)
arr
```

```python
arr.dtype
```

```python
# 기존의 타입을 변경(astype())
float_arr = arr.astype(np.float64)
float_arr
```

# array 관련 연산
두 array 간에는 더하기, 빼기, 곱하기, 나누기 등의 연산을 수행할 수 있다. 동일한 위치의 성분끼리 이루어지게 되므로, 당연히 두 array의 모양이 같아야 계산이 가능하다

```python
arr1 = np.array([[1,2,3],[4,5,6]], dtype=np.float64)
arr2 = np.array([[7,8,9],[10,11,12]], dtype=np.float64)
```

```python
arr1
```

```python
arr2
```

```python
# 2차원에서 같은 위치의 성분끼리 연산이 된다.
arr1 + arr2
```

```python
arr1 * arr2
```

```python
arr1 ** arr2
```

```python
1 / arr1
```
